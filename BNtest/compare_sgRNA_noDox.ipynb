```python
import pandas as pd
```

```python
# --- input Table S5: Result in gene turn-off experiments
xlsx_path = r"Table S5 Comparison of Fitness between Murine Pneumonia Model and C+Y Medium, Related to Figures 4 and 5.xlsx"
sheet_name = 0
dfS5 = pd.read_excel(xlsx_path, sheet_name=sheet_name, engine="openpyxl")

# --- filter: interaction in-vitro & in-vivo = "yes"
mask = (dfS5["interaction"].astype(str).str.strip().str.lower().eq("yes"))
sgrna_series = dfS5.loc[mask, "sgRNA"].dropna()

interaction_sgrnas = sgrna_series.astype(str).unique().tolist()
print(f"sgRNAs have interaction between in-vitro and in-vivo: {len(interaction_sgrnas)}")
#pd.Series(interaction_sgrnas, name="sgRNA").to_csv("sgRNAs_interaction_yes_unique.csv", index=False)
```

```python
# Input file (as uploaded)
df = pd.read_csv("neutrality_noDox_FDR0.05.csv")
sgrnas_non_neutral = df["sgRNA"].dropna().tolist()
#pd.Series(sgrnas_non-neutral, name="sgRNA").to_csv("sgRNAs.csv", index=False)
print(f"Extracted {len(sgrnas_non_neutral)} sgRNAs")
```

```python
joint = sorted(set(interaction_sgrnas).intersection(sgrnas_non_neutral))
print(joint)
```
