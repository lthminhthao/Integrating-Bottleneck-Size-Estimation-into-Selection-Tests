### Comparison with in-vivo data in Table S5

```python
# --- input Table S5: Result in gene turn-off experiments
xlsx_path = r"Table S5 Comparison of Fitness between Murine Pneumonia Model and C+Y Medium, Related to Figures 4 and 5.xlsx"
sheet_name = 0
dfS5 = pd.read_excel(xlsx_path, sheet_name=sheet_name, engine="openpyxl")

# --- filter: mice in-vivo = "essential"
mask = (dfS5["Mice.essentiality"].astype(str).str.strip().str.lower().eq("essential"))
sgrna_series = dfS5.loc[mask, "sgRNA"].dropna()

essential_invivo_sgrnas = sgrna_series.astype(str).unique().tolist()
print(f"essential sgRNAs in-vivo: {len(essential_invivo_sgrnas)}")
```

```python
# Input file neutrality test Dox
df = pd.read_csv("neutrality_test_24hpi_Dox.csv")
mask = (df["direction"].astype(str).str.strip().str.lower().eq("down_in_recipient"))
sgrna_series1 = df.loc[mask, "feature"].dropna()

essential_Nb_sgrnas = sgrna_series1.astype(str).unique().tolist()
#essential_Nb_sgrnas = [re.sub(r'^(sgRNA)(\d+)$',lambda m: f"{m.group(1)}{int(m.group(2)):04d}", s) for s in essential_Nb_sgrnas]
print(f"essential sgRNAs neutrality test: {len(essential_Nb_sgrnas)}")
```

```python
joint = sorted(set(essential_invivo_sgrnas).intersection(essential_Nb_sgrnas))
print(len(joint))
print("-------------------------------")
print(joint)
```
